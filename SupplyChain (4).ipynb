```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
```

```python
df["Unit_Price"] = df["Unit_Price"].fillna(df["Unit_Price"].mean())
```

```python
df["Supplier"] = df["Supplier"].fillna("Unknown")
```

```python
df["Warehouse"] = df["Warehouse"].fillna(df["Warehouse"].mode()[0])
```

```python
df["Delivery_Date"] = df["Delivery_Date"].fillna(df["Order_Date"])
```

```python
df["Total_Amount"] = df["Total_Amount"].fillna(
    df["Quantity"] * df["Unit_Price"]
)
```

```python
df.isnull().sum()
```

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df["Product_Name"] = df["Product_Name"].str.title()
df["Supplier"] = df["Supplier"].str.title()
df["Transport_Mode"] = df["Transport_Mode"].str.title()
df["Delivery_Status"] = df["Delivery_Status"].str.upper()
df["Payment_Status"] = df["Payment_Status"].str.upper()
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Delivery_Date"] = pd.to_datetime(df["Delivery_Date"])
```

```python
df.rename(columns={
    "Unit_Price": "Price_Per_Unit",
    "Employee_ID": "Manager_ID"
}, inplace=True)
```
